```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Assignment3_result.csv")
df.head()
```

```python
plt.figure(figsize=(10, 6))

# Plot each ordering type
for ordering in df["ordering"].unique():
    subset = df[df["ordering"] == ordering]
    plt.plot(subset["n"], subset["time(ms)"], marker='o', label=ordering)

# Formatting the plot
plt.xlabel("Input Size (n)")
plt.ylabel("Time (ms)")
plt.title("Insertion Sort Growth Rate for Different Orderings")
plt.legend()
plt.grid(True)

# Show plot
plt.show()
```

Based on our understanding of the insertion sort, we know that the ordered array should take least steps and least time; the reverse-order array is the worst case; the randomly-ordered one should be in between these two, while the partially-order in between the random and the best case (cause we set it to be half ordered, half random for simplicity). This plot confirms this order of complexity for insertion sort we know across these four kinds of integer arrays.


For the reverse-ordered array, each iteration compares adjacent elements and swap. Every element must move all previous elements before inserting itself, meaning that each element swaps i times where i is its position. So the total number of swaps is O(n^2) with each swap gives at least 2 hits of the array elements. The plot confirms our knowlege that it takes most time, with a curve showing the shape of a quadratic growth.

For the ordered array, each iteration only compares adjacent elements but never swaps (saving at least 2 "hits" of the array elements in swap for each iteration). As each element is compared once and break the nested loop, we have O(n) comparisons with 0 swaps. The plot confirms our knowlege that it takes least time, while it's too close to the x axis, it's hard to read the details.

For the randomly-ordered array, it's something in between the first two cases, and on average we can think of each elements needs to swaps half the elements before it, while it's still O(n^2) neglecting the coefficients. The plot confirms our knowlege that its curve lays in between the first two (with a slope in between the first two too), with a curve still shows some shape of a quadratic growth.

For the partially-ordered array, it's something in between the last two cases (ordered and randomly-ordered), because we set it to be half ordered and half random for simplicity. The plot confirms our knowlege that its curve lays in between the last two (with a slope in between the first two too), with a curve still shows some shape of a quadratic growth.
