```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("Assignment3_result.csv")
data.head()
```

```python
# header: n(#_of_runs),m(#_of_Steps),d(AvgDist)
# m_Array = {10, 20, 50, 100, 200, 500, 1000, 2000, 5000}

plt.figure(figsize=(8, 5))
plt.plot(data["m(#_of_Steps)"], data["d(AvgDist)"], marker='o', linestyle='-')
plt.title("Relationship Between the Number of Steps (m) and Average Distance (d) (200 Iterations for Each m)")
plt.xlabel("Number of Steps (m)")
plt.ylabel("Average Distance (d)")
plt.grid()
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5)) 

ax1.plot(data["m(#_of_Steps)"], data["d(AvgDist)"], marker='o', linestyle='-')
ax1.set_title("Full Range: Steps (m) vs. Average Distance (d)")
ax1.set_xlabel("Number of Steps (m)")
ax1.set_ylabel("Average Distance (d)")
ax1.grid()


zoom_in_data = data.head(5)  
ax2.plot(zoom_in_data["m(#_of_Steps)"], zoom_in_data["d(AvgDist)"], marker='o', linestyle='-')
ax2.set_title("Zoomed-In: First 5 Steps (m) vs. Average Distance (d)")
ax2.set_xlabel("Number of Steps (m)")
ax2.set_ylabel("Average Distance (d)")
ax2.grid()

plt.tight_layout()
plt.show()
```

These two plots both looks like the curve for a square root graph showing positive relation, where the avg distance d increases as m increases while curving under the x=y line. To further estimate the relation between these two, we'll assume there is a square-root growth, and perform a linear regression using d as the target feature and the square root of m as the independent feature.

```python
import numpy as np
from sklearn.linear_model import LinearRegression



m = data["m(#_of_Steps)"].values.reshape(-1, 1)  
d = data["d(AvgDist)"].values  


model = LinearRegression()
model.fit(np.sqrt(m), d)
d_pred = model.predict(np.sqrt(m))


print(f"Coefficient: {model.coef_[0]}")
print(f"Intercept: {model.intercept_}")


plt.figure(figsize=(8, 5))
plt.scatter(np.sqrt(m), d, color="blue", label="Actual Data")
plt.plot(np.sqrt(m), d_pred, color="red", label="Regression Line")
plt.title("Regression: d vs. sqrt(m)")
plt.xlabel("sqrt(m)")
plt.ylabel("Average Distance (d)")
plt.legend()
plt.grid()
plt.show()
```

```python
r_square_score = model.score(np.sqrt(m), d)
print(r_square_score)
```

As we can see in the graph, we have a line capsture most of the variaty of our d (avg distance) value, with the r^2 almost 1. We can say it with confidence that the relation d = 0.887* sqrt(m) + 0.193 between the numeber of steps (m) and the final distance to starting point (d) in our RandomWalk experiment is very close to the true relation between d and m in RandomWalk problem.
